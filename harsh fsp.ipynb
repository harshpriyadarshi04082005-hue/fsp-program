# Write a Python program to create a DataFrame containing missing values. Perform the following operations

```python
import pandas as pd
import numpy as np


data = {
    'A': [10, np.nan, 30, 40, np.nan],
    'B': [5, 15, np.nan, 35, 50]
}
df = pd.DataFrame(data)
print("Original DataFrame:\n", df, "\n")


print("Missing Value Detection (True indicates missing):\n", df.isnull(), "\n")

df_filled = df.copy()
df_filled['A'] = df_filled['A'].fillna(df_filled['A'].mean())    # Filling with Mean
df_filled['B'] = df_filled['B'].fillna(df_filled['B'].median())  # Filling with Median
print("DataFrame after replacing missing values:\n", df_filled, "\n")


df_dropped = df.dropna()
print("DataFrame after dropping rows with missing values:\n", df_dropped)
```

# 2. Write a Python program using Pandas to: 
• Create a dataset containing duplicate rows  
• Identify duplicate entries  
• Remove all duplicate rows from the dataset 

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'David'],
    'Age': [25, 30, 25, 35, 30, 40],
    'City': ['New York', 'London', 'New York', 'Paris', 'London', 'Tokyo']
}

df = pd.DataFrame(data)
print("--- Original DataFrame ---")
print(df)
print("\n" + "="*40 + "\n")


duplicate_mask = df.duplicated()
print("--- Duplicate Rows Identification (True indicates a duplicate) ---")
print(duplicate_mask)


print("\nActual duplicate rows:")
print(df[duplicate_mask])
print("\n" + "="*40 + "\n")


df_cleaned = df.drop_duplicates()

print("--- DataFrame After Removing Duplicates ---")
print(df_cleaned)
```

# 3. Write a Python program to normalize the following list of values using Min-Max 
normalization: 
[12, 25, 40, 55, 70] 
• Use the formula: 
• 𝑥′ = 𝑥−𝑥𝑚𝑖𝑛/
𝑥𝑚𝑎𝑥−𝑥𝑚𝑖�

```python
import numpy as np


values = np.array([12, 25, 40, 55, 70])


x_min = values.min()
x_max = values.max()


normalized_values = (values - x_min) / (x_max - x_min)


print("--- Min-Max Normalization Results ---")
print(f"Original Values: {values.tolist()}")
print(f"Minimum Value (x_min): {x_min}")
print(f"Maximum Value (x_max): {x_max}")
print("-" * 37)
print(f"Normalized Values: {normalized_values.tolist()}")


print("\nStep-by-step Transformation:")
for orig, norm in zip(values, normalized_values):
    print(f"Original: {orig:2d}  -->  Normalized: {norm:.4f}")
```

# . Write a Python program to standardize a dataset using Z-score normalization. 
The program should: 
• Calculate mean and standard deviation  
• Compute Z-score for each value  
• Display transformed data 

```python
import numpy as np


dataset = np.array([15, 25, 35, 45, 55])
print("--- Original Dataset ---")
print(dataset)
print("\n" + "="*40 + "\n")

mean_val = np.mean(dataset)
std_dev_val = np.std(dataset)

print("--- Statistical Calculations ---")
print(f"Calculated Mean (μ): {mean_val}")
print(f"Calculated Standard Deviation (σ): {std_dev_val:.4f}")
print("\n" + "="*40 + "\n")

z_scores = (dataset - mean_val) / std_dev_val


print("--- Transformed Data (Z-score Normalized) ---")
print(z_scores)

print("\nDetailed breakdown per element:")
for original, z in zip(dataset, z_scores):
    print(f"Original Value: {original:2d}  --->  Z-score: {z:.4f}")
```

# Write a Python program to detect outliers in a dataset using the Z-score method. 

```python
import numpy as np


dataset = np.array([10, 12, 12, 13, 12, 11, 14, 13, 950, 10, 15, 12, 14, -400, 11, 13])

print("--- Original Dataset ---")
print(dataset)
print(f"Total data points: {len(dataset)}\n")


mean = np.mean(dataset)
std_dev = np.std(dataset)

z_scores = (dataset - mean) / std_dev


threshold = 3


outliers = dataset[np.abs(z_scores) > threshold]
clean_data = dataset[np.abs(z_scores) <= threshold]

print("--- Z-Score Outlier Detection Results ---")
print(f"Calculated Mean: {mean:.2f}")
print(f"Calculated Std Dev: {std_dev:.2f}\n")

print(f"Detected Outliers: {outliers}")
print(f"Cleaned Dataset (Outliers removed):")
print(clean_data)


print("\nDetailed Outlier Breakdown:")
for val, z in zip(dataset, z_scores):
    if np.abs(z) > threshold:
        print(f"-> VALUE: {val:4d} has a Z-Score of {z:6.2f} (EXCEEDS ±{threshold})")
```

#  Write a Python program to: 
• Create a dataset containing categorical values such as ["Red", "Blue", "Green", "Blue"]  
• Convert the categorical data into numerical labels using Label Encoding 


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder


data = {
    'Color': ["Red", "Blue", "Green", "Blue"]
}
df = pd.DataFrame(data)

print("--- Original DataFrame ---")
print(df)
print("\n" + "="*40 + "\n")


encoder = LabelEncoder()


df['Color_Encoded'] = encoder.fit_transform(df['Color'])

print("--- DataFrame After Label Encoding ---")
print(df)
print("\n" + "="*40 + "\n")


print("--- Encoder Mapping Summary ---")
for index, label in enumerate(encoder.classes_):
    print(f"Text Label: {label:5s}  ---> Assigned Integer: {index}")
```

# Write a Python program to perform One-Hot Encoding on a categorical feature named City 
containing values:

# . Write a Python program using StandardScaler from Scikit-learn to scale numerical features of a dataset. 

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler


data = {
    'Age': [25, 35, 45, 20, 50],
    'Estimated_Income': [50000, 85000, 120000, 30000, 105000]
}
df = pd.DataFrame(data)

print("--- Original DataFrame ---")
print(df)
print("\n" + "="*45 + "\n")


scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)


df_scaled = pd.DataFrame(scaled_data, columns=df.columns)

print("--- Standard Scaled DataFrame ---")
print(df_scaled)
print("\n" + "="*45 + "\n")

print("--- Verification Metrics ---")
print("Means (Should be close to 0):")
print(df_scaled.mean().round(4))
print("\nStandard Deviations (Should be exactly 1):")
print(df_scaled.std(ddof=0).round(4)) # ddof=0 matches population std dev used by StandardScaler
```

# Write a Python program to compare: 
• Min-Max Scaling  
• Standardization  
on the same dataset and display the transformed outputs. 

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

data = {
    'Original_Data': [10, 20, 35, 50, 90]
}
df = pd.DataFrame(data)


min_max_scaler = MinMaxScaler()
df['Min-Max_Scaled'] = min_max_scaler.fit_transform(df[['Original_Data']])


std_scaler = StandardScaler()
df['Standardized'] = std_scaler.fit_transform(df[['Original_Data']])

print("--- Scaling Comparison Output ---")
print(df.to_string(index=False))

print("\n" + "="*50 + "\n")


print("--- Summary of Key Differences Observed ---")
print(f"Original Range:  {df['Original_Data'].min()} to {df['Original_Data'].max()}")
print(f"Min-Max Range:   {df['Min-Max_Scaled'].min():.1f} to {df['Min-Max_Scaled'].max():.1f} (Strictly bound between 0 and 1)")
print(f"Standard Mean:   {df['Standardized'].mean().round(1)} (Centered precisely at 0)")
# Fixed by wrapping the calculation inside round()
print(f"Standard Deviation: {round(df['Standardized'].std(ddof=0), 1)} (Rescaled to a unit variance of 1)")
```

# 10. Write a Python program to convert a column containing numeric strings into integer type. 

```python
import pandas as pd


data = {
    'Item': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Quantity_String': ['15', '42', '28', '10']
}
df = pd.DataFrame(data)

print("--- Before Conversion ---")
print(df)
print("\nData Types:")
print(df.dtypes)  # 'object' type in Pandas usually indicates strings
print("\n" + "="*40 + "\n")


df['Quantity_Int'] = df['Quantity_String'].astype(int)

print("--- After Conversion ---")
print(df)
print("\nData Types:")
print(df.dtypes)  # Notice 'Quantity_Int' is now int64
```
